# 소득분위 계층별 소비구조 추세 분석
- **용도:** 2026 국가데이터활용대회 공모전 참가 및 학생설계융합전공 졸업논문
  
- **분석 주제:** 2022Q1~2026Q1(17개 분기) 동안 **소득 5분위 계층별 소비구조(12대 항목 비중)** 가 어떻게 변했고, 계층 간 격차가 어떻게 움직였는지 추세 분석
  
- **분석 활용 데이터:** KOSIS(품목별/지출목적별 소비자물가지수), MDIS(가계동향조사)

- **문제의식:** 체감 물가와 생활 압박은 누구에게 더 무겁게 작용했는가, 각 계층은 무엇을 줄이고 무엇으로 옮기며 적응했는가, 그리고 계층 간 소비구조의 격차는 좁혀졌는가 벌어졌는가. 우리가 흔히 접하는 '소비자물가 상승률'이나 '가구당 평균 소비지출'은 평균적인 한 가구를 묘사하지만, 현실은 같은 물가와 같은 경기 충격에도 계층마다 전혀 다르게 닿는다. 따라서 단일 평균 통계로는 보이지 않는 이 분배적 실상을 드러내는 것이 분석의 출발점이다.

- **이 분석이 기여할 수 있는 부분**

    - **1. 정책의 정밀화 — 평균이 아닌 분포로.** 가구동향과 CPI를 분위×분기×항목으로 결합해 계층별 체감물가(각 분위의 소비바스켓 × 항목별 물가)를 산출하면, "어느 계층이 실제로 가장 높은 물가 부담을 졌는가"를 정량화할 수 있습니다. 이는 에너지·식료품 바우처, 타깃 복지 같은 민생 지원을 체감 부담이 큰 곳에 배분하는 근거가 됩니다. 통계청도 같은 문제의식에서 가구특성별 물가지수(실험적통계)를 만들었으니, 우리 결과는 그 공식 지표와 교차검증되며 보완할 수 있습니다.

    - **2. 소비 위축의 조기 진단.** 어떤 계층이 어떤 항목을 줄여가는지를 시계열로 추적하면, 위축이 건강한 조정인지 고통스러운 긴축인지 구분할 수 있습니다. 저소득층이 보건·교육 지출을 줄이는 신호가 보인다면 이는 선제적 개입이 필요한 경보입니다.

    - **3. 방법론적 기여.** 가계동향(신분류)과 CPI(구분류)는 그대로는 맞지 않아, 품목 단위 재집계로 정보통신·오락문화를 신분류에 맞춘 분류 정합 디플레이터를 만들었습니다. 이로써 명목과 실질을 분리해 "값이 올라 더 쓴 것"과 "실제로 더(덜) 소비한 것"을 가를 수 있습니다 — 후생 해석의 핵심이며, 다른 분석에도 재사용 가능한 자산입니다.

    - **4. 불평등 논의의 실증 근거.** 엥겔계수(식료품 비중)·슈바베계수(주거비 비중)의 계층별 추이는 생활수준·주거부담의 분배적 변화를 수치로 보여주어, 소비 양극화 논의에 데이터 기반 근거를 제공합니다. 기업 입장에서도 계층별 소비구조 변화는 수요예측·세그먼트 전략의 입력이 됩니다.

- **데이터의 한계점:**
이 데이터는 **반복 횡단면**이라 같은 가구를 추적하는 패널이 아니므로, 개별 가구의 행동 적응보다 집단 평균의 구조 변화를 봅니다. 17개 분기로는 인과 규명보다 **기술·진단**에 강하며, 비중·체감물가는 상관적·기술적 증거이지 인과가 아닙니다. 또한 여기서 '계층'은 **소득** 기준이라 자산·부채 차원은 별도로 보아야 합니다. 이 경계를 분명히 둔 채 해석하면, 위의 기여들이 과장 없이 견고해집니다.
**요약하면, 이 분석은 "평균이 가린 계층별 소비 현실과 물가 부담의 불균형"에서 출발해, 누가 무엇에 얼마나 눌렸는지를 분배적으로 드러내고, 그 진단을 정책·시장이 정밀하게 활용하도록 돕는 데 기여합니다.**

- **데이터 기반 문제 정의:** 2022Q1~2026Q1(17개 분기) 가중평균 지출을 이용해, ① 12대 소비항목의 지출 비중(소비구조) 변화 추세를 정량화하고, ② 엥겔계수·슈바베계수 등 후생 지표의 추이를 추적하며, ③ 경기·물가에 민감하게 반응하는 민감 항목을 식별하고, ④ 향후 분기 소비를 전망하는 문제로 정의합니다.

|-----------|--------|---------|
| 소비구조 변화 추세 정량화 | 항목별 비중 추이·증감률 | 위축/팽창 항목 조기 포착 → 선제적 민생 대응 |

| 분석 목표 | 산출물 | 기대효과 |
|---|---|---|
| 계층별 소비구조 변화 | 분위×항목 비중 추이 | 위축/팽창 항목 조기 포착 |
| 엥겔·슈바베계수 추이 | 후생 지표 시계열 | 체감 생활비·주거부담 진단 |
| 계층 간 격차 | 격차지표 추세 | 소비 양극화 모니터링 |
| 계층별 체감물가 | 분위 바스켓×항목 CPI | 물가의 차등 부담 정량화 |

아래에서 데이터 규모를 먼저 확인합니다.

```python
files = sorted(glob.glob(os.path.join(UPLOAD_DIR, "2*_분기자료*.csv")))
rows = []
for f in files:
    d = pd.read_csv(f, encoding="cp949", usecols=["조사연월"])
    rows.append({"연도": os.path.basename(f)[:4], "행수": len(d),
                 "분기수": d["조사연월"].nunique()})
overview = pd.DataFrame(rows)
print(overview.to_string(index=False))
print("총 행수:", overview["행수"].sum())
```

## 2-c. CPI 통합 (지출목적별) → `cpi_long`

명목지출을 실질로 환산하기 위해 지출목적별 CPI(2020=100)를 결합합니다. **CPI는 구분류(COICOP-K 2008: 08 통신, 09 오락문화)** 라, 신분류를 쓰는 가계동향과 **정보통신·오락문화 2개 항목만** 어긋납니다(나머지 10개는 직접 매칭). 통계정보보고서가 *"2025년 기준 개편 시 재분류 예정"* 으로 확정.

```python
def q_label(c):
    y, q = str(c).split("."); return f"{y}Q{q.split('/')[0]}"

fp = glob.glob(os.path.join(UPLOAD_DIR, "지출목적별_소비자물가지수*"))[0]
raw = pd.read_excel(fp, sheet_name="데이터", header=0)
raw.columns = ["시도별", "지출목적별"] + list(raw.columns[2:])
raw = raw.drop(columns=["시도별"])
raw = raw.rename(columns={c: q_label(c) for c in raw.columns if c != "지출목적별"})
raw["항목"] = raw["지출목적별"].astype(str).str.strip().map(CPI_DIV_MAP)
cpi_long = (raw[raw["항목"].notna()].drop(columns="지출목적별")
            .melt(id_vars="항목", var_name="분기라벨", value_name="CPI"))
cpi_long.to_pickle(os.path.join(WORK_DIR, "cpi_long.pkl"))

piv = cpi_long.pivot(index="항목", columns="분기라벨", values="CPI")
print("[2022Q1→2026Q1 누적 물가상승률(%) · 구분류 기준]")
print(((piv["2026Q1"]/piv["2022Q1"]-1)*100).round(2).sort_values(ascending=False).to_string())
```

## 2-d. 품목 단위 재집계 디플레이터

CPI의 큰 분류(08·09)를 그대로 쓰지 않고, **품목별 CPI를 신분류 정의에 맞게 재조립**합니다.
- **정보통신** = 통신(휴대전화기·유선/휴대전화료·인터넷) + 영상·IT 블록(TV·영상음향·컴퓨터·방송수신료·OTT…)
- **오락·문화** = CPI 09에서 영상·IT 블록을 제외한 나머지(단체여행·운동오락·문화서비스·게임장난감·서적·반려동물·문구…)

각 하위항목 가격 = 대응 CPI 품목의 평균, 가중치 = 가계동향 하위지출(전기간 가중평균). 우편료는 신분류상 교통·운송이므로 정보통신에서 제외.

> 아래 셀은 **품목별 CPI 파일이 있으면 원본에서 재집계**하고, 없으면 직전에 저장한 `cpi_long_method1.pkl`을 사용합니다.

```python
def _weighted_mean(frame, col): return np.average(frame[col], weights=frame["가중값"])

item_fp = glob.glob(os.path.join(UPLOAD_DIR, "품목별_소비자물가지수*"))
if item_fp:
    icpi = pd.read_excel(item_fp[0], sheet_name="데이터", header=0)
    icpi.columns = ["시도", "품목"] + list(icpi.columns[2:])
    icpi = icpi.drop(columns=["시도"]).dropna(subset=["품목"]).reset_index(drop=True)
    icpi["name"] = (icpi["품목"].astype(str).str.replace("　","",regex=False)
                    .str.replace(" ","",regex=False).str.strip("()"))
    qc = [c for c in icpi.columns if c not in ("품목","name")]
    icpi = icpi.rename(columns={c: q_label(c) for c in qc})
    QL = [q_label(c) for c in qc]
    cidx = icpi.set_index("name")[QL]

    def sub_price(items):
        if not items: return None
        return pd.concat([cidx.loc[n].astype(float) for n in items], axis=1).mean(axis=1)

    def composite(mapping, cat):
        s, wt = {}, {}
        for sub, items in mapping.items():
            sp = sub_price(items)
            if sp is None: continue
            s[sub] = sp; wt[sub] = _weighted_mean(df, f"{PRE}{cat}_{sub}")
        wser = pd.Series(wt); wser /= wser.sum()
        return pd.DataFrame(s).T.mul(wser, axis=0).sum(axis=0), wser

    ict_def, ict_w = composite(ICT_MAP, "정보통신")
    rec_def, rec_w = composite(REC_MAP, "오락문화")
    cl = pd.read_pickle(os.path.join(WORK_DIR,"cpi_long.pkl")).pivot(
        index="항목", columns="분기라벨", values="CPI")
    cl.loc["정보통신"] = ict_def; cl.loc["오락·문화"] = rec_def
    cl.reset_index().melt(id_vars="항목", var_name="분기라벨", value_name="CPI"
        ).to_pickle(os.path.join(WORK_DIR,"cpi_long_method1.pkl"))
    print("[정보통신 하위 가중치(%)]"); print((ict_w*100).round(1).to_string())
    print("\n품목별 CPI 원본에서 방법1 재집계 완료.")
else:
    print("품목별 CPI 파일 없음 → 저장된 cpi_long_method1.pkl 사용 (UPLOAD_DIR에 파일을 두면 원본 재집계).")
```

## 3. 데이터 전처리 — 분기×소득5분위 패널 구성

가구 단위 `df_raw`를 **`분기 × 분위`(17×5=85행) 가중집계 패널**로 변환한다.
- 비중(명목): 각 항목 ÷ 소비지출금액 (분기 내 합=1) → 소비구조
- 실질지출: 방법1 디플레이터로 명목→실질 환산(가격 vs 소비량 분리)
- 엥겔계수(식료품 비중)·슈바베계수(주거비 비중): 후생·생활부담 지표
- 계층별 체감물가: 분위별 2022Q1 바스켓 고정(라스파이레스) × 항목 CPI

```python
import os, numpy as np, pandas as pd

# 이전 셀에서 정의된 CAT12·PRE·WORK_DIR 사용 (안전하게 산출물 재로딩)
df   = pd.read_pickle(os.path.join(WORK_DIR, "df_raw.pkl"))
defl = pd.read_pickle(os.path.join(WORK_DIR, "cpi_long_method1.pkl")) \
         .pivot(index="항목", columns="분기라벨", values="CPI").T   # index=분기라벨, col=항목
CATS = list(CAT12.keys()); w = "가중값"

# --- 분기 × 분위 가중집계 ---
def wavg(g, c): return np.average(g[c], weights=g[w])
rows = []
for (ql, q), g in df.groupby(["분기라벨", "분위"]):
    rec = {"분기라벨": ql, "분위": int(q), "총소비_명목": wavg(g, "가계지출_소비지출금액")}
    for k, c in CAT12.items():
        rec[f"명목_{k}"] = wavg(g, c)
    rows.append(rec)
panel = pd.DataFrame(rows)
panel["연도"] = panel["분기라벨"].str[:4].astype(int)
panel["분기"] = panel["분기라벨"].str[-1].astype(int)
QORDER = sorted(panel["분기라벨"].unique())
panel["t"] = panel["분기라벨"].map({q: i for i, q in enumerate(QORDER)})

# --- 비중 / 실질 / 엥겔·슈바베 ---
for k in CATS:
    panel[f"비중_{k}"] = panel[f"명목_{k}"] / panel["총소비_명목"]
    panel[f"실질_{k}"] = panel.apply(lambda r: r[f"명목_{k}"] / (defl.loc[r["분기라벨"], k] / 100), axis=1)
panel["총소비_실질"] = panel[[f"실질_{k}" for k in CATS]].sum(axis=1)
panel["엥겔계수"]  = panel["비중_식료품·비주류음료"]
panel["슈바베계수"] = panel["비중_주거·수도·광열"]

# --- 계층별 체감물가 (2022Q1 분위별 비중 고정 = 라스파이레스) ---
base = panel[panel["분기라벨"] == "2022Q1"].set_index("분위")
def personal_cpi(r):
    bw = base.loc[r["분위"], [f"비중_{k}" for k in CATS]].values
    p  = defl.loc[r["분기라벨"], CATS].values
    return float(np.dot(bw, p))
panel["체감물가"] = panel.apply(personal_cpi, axis=1)

panel = panel.sort_values(["분위", "t"]).reset_index(drop=True)
panel.to_pickle(os.path.join(WORK_DIR, "panel.pkl"))
print("panel:", panel.shape)
print("\n[체감물가 누적상승률 2022Q1→2026Q1, 분위별 %]  (역진성 점검)")
for q in range(1, 6):
    s = panel[panel["분위"] == q].set_index("분기라벨")["체감물가"]
    print(f"  분위{q}: {(s['2026Q1']/s['2022Q1']-1)*100:5.2f}%")
```

### 3-a. 계층 간 소비구조 격차 지표

- 항목별 격차 = (분위5 비중 − 분위1 비중), 단위 %p
- 구조격차지수 = ½·Σ|분위5 비중 − 분위1 비중|  (0=동일 구조, 1=완전 상이)

```python
share = panel.pivot(index="분기라벨", columns="분위", values=[f"비중_{k}" for k in CATS])
gap = pd.DataFrame(index=QORDER)
for k in CATS:
    gap[k] = (share[f"비중_{k}"][5] - share[f"비중_{k}"][1]) * 100   # 분위5-분위1 (%p)
diss = {}
for ql in QORDER:
    w1 = np.array([panel[(panel.분위==1)&(panel.분기라벨==ql)][f"비중_{k}"].values[0] for k in CATS])
    w5 = np.array([panel[(panel.분위==5)&(panel.분기라벨==ql)][f"비중_{k}"].values[0] for k in CATS])
    diss[ql] = 0.5 * np.abs(w5 - w1).sum()
gap["구조격차지수"] = pd.Series(diss)

print("[격차: 분위5-분위1 (%p) + 구조격차지수]")
print(gap.loc[["2022Q1","2024Q1","2026Q1"],
              ["식료품·비주류음료","교육","교통·운송","오락·문화","구조격차지수"]].round(3).to_string())
```

### 3-b. 입력(X)·출력(Y) 정의 및 Train/Validation 분할

소비구조(비중) 추세를 학습·예측하기 위한 지도학습 데이터를 구성한다.
- Y: 비중(항목별)  /  X: 시점 t, 분기(계절) 더미, 분위 더미, 항목 더미
- 시계열 누수 방지를 위해 **시점 기준 분할** — 검증=최근 3개 분기(2025Q3~2026Q1)
- 산출물: X_train, X_val, Y_train, Y_val

```python
long = panel.melt(id_vars=["분기라벨","분위","t","분기"],
                  value_vars=[f"비중_{k}" for k in CATS],
                  var_name="항목", value_name="비중")
long["항목"] = long["항목"].str[3:]

Xall = pd.get_dummies(long[["t","분기","분위","항목"]].astype({"분기":str,"분위":str}),
                      columns=["분기","분위","항목"], drop_first=True).astype(float)
Yall = long["비중"].astype(float)

val_q = ["2025Q3","2025Q4","2026Q1"]
is_val = long["분기라벨"].isin(val_q).values
X_train, X_val = Xall[~is_val], Xall[is_val]
Y_train, Y_val = Yall[~is_val], Yall[is_val]
for nm, obj in [("X_train",X_train),("X_val",X_val),("Y_train",Y_train),("Y_val",Y_val)]:
    obj.to_pickle(os.path.join(WORK_DIR, f"{nm}.pkl"))
print(f"Train {X_train.shape} / Val {X_val.shape}  (검증=2025Q3~2026Q1)")
```

### 3-c. (참고) 수요·탄력성 회귀 — 계층별 해석

**Working-Leser 예산몫 모형**으로 분위별 **지출(엥겔) 탄력성** e를 도출한다.
e_i = 1 + β_i / w̄_i  (e<1 필수재, e>1 사치재)

추가로 **"식료품 가격 1%↑ 시 비식료품 실질지출 반응(η)"** 을 1차차분으로 추정한다.
단, CPI 물가는 전 분위 공통(단일 국가지수)이고 표본이 17분기뿐이라
**가격 반응(η)은 식별이 약한 탐색적·기술적 결과**이며 인과로 해석하지 않는다.
(엄밀한 가격탄력성은 향후 AIDS 수요체계로 확장)

> **통계적 처리.** 표준오차는 HAC(Newey-West, 2시차)로 분기 자기상관을 보정하고 95% 신뢰구간을 함께 보고한다.
> Working-Leser에서 설명변수 ln(실질총소비)는 종속변수(비중)의 분모와 기계적으로 연결되는 알려진 내생성이 있으므로,
> 계수는 인과가 아닌 기술적 연관으로 해석한다.


```python
# 필요 시: pip install statsmodels
import statsmodels.formula.api as smf

panel["ln_real_total"] = np.log(panel["총소비_실질"])

HAC = dict(cov_type="HAC", cov_kwds={"maxlags": 2})  # Newey-West: 분기 자기상관 보정

# (1) 지출(엥겔) 탄력성 by 분위 — 점추정 + 95% 신뢰구간
def expend_elasticity(cat):
    d = panel.copy(); d["w"] = d[f"비중_{cat}"]; d["분위"] = d["분위"].astype(str)
    m = smf.ols("w ~ ln_real_total * C(분위) + C(분기)", data=d).fit(**HAC)
    V = m.cov_params()
    out = {}
    for q in map(str, range(1, 6)):
        key = f"ln_real_total:C(분위)[T.{q}]"
        b = m.params["ln_real_total"] + m.params.get(key, 0.0)
        var = V.loc["ln_real_total", "ln_real_total"]
        if key in m.params.index:
            var += V.loc[key, key] + 2 * V.loc["ln_real_total", key]
        se = float(var) ** 0.5
        wbar = d[d["분위"] == q]["w"].mean()   # w̄ 고정 가정의 근사 CI
        out[q] = (1 + b / wbar, 1 + (b - 1.96 * se) / wbar, 1 + (b + 1.96 * se) / wbar)
    return out

print("[지출(엥겔) 탄력성 e_i by 분위 — 점추정 [95% CI], HAC]  e<1 필수재 / e>1 사치재")
print("※ CI가 1을 포함하면 필수재/사치재 구분이 통계적으로 불확실")
for cat in ["식료품·비주류음료", "교육", "오락·문화"]:
    es = expend_elasticity(cat)
    print(f"  {cat}")
    for q, (e, lo, hi) in es.items():
        print(f"    분위{q}: {e:5.2f}  [{lo:5.2f}, {hi:5.2f}]")

# (2) 식료품가격↑ → 비식료품 실질지출 반응 η (1차차분, 탐색적) — 95% CI 포함
panel["ln_real_nonfood"] = np.log(panel["총소비_실질"] - panel["실질_식료품·비주류음료"])
panel["ln_pfood"] = panel["분기라벨"].map(np.log(defl["식료품·비주류음료"]))
panel = panel.sort_values(["분위", "t"])
panel["d_nonfood"] = panel.groupby("분위")["ln_real_nonfood"].diff()
panel["d_pfood"]   = panel.groupby("분위")["ln_pfood"].diff()
d = panel.dropna(subset=["d_nonfood", "d_pfood"]).copy(); d["분위"] = d["분위"].astype(str)
m = smf.ols("d_nonfood ~ d_pfood * C(분위) + C(분기)", data=d).fit(**HAC)
V = m.cov_params()
print("\n[식료품가격 1%↑ 시 비식료품 실질지출 반응 η — 점추정 [95% CI], 탐색적]")
for q in map(str, range(1, 6)):
    key = f"d_pfood:C(분위)[T.{q}]"
    eta = m.params["d_pfood"] + m.params.get(key, 0.0)
    var = V.loc["d_pfood", "d_pfood"]
    if key in m.params.index:
        var += V.loc[key, key] + 2 * V.loc["d_pfood", key]
    se = float(var) ** 0.5
    print(f"  분위{q}: {eta:+.2f}%  [{eta - 1.96 * se:+.2f}, {eta + 1.96 * se:+.2f}]")
```

### 3-d. 2026년 2~3분기 소비구조 단기 예측

분위×항목별 비중을 **추세(t) + 분기 계절더미** 로 적합해 2026Q2·Q3를 전망하고,
분위 내 합이 1이 되도록 재정규화한다. (17개 분기 기반의 기술적 단기 전망)
검증: 최근 3분기 홀드아웃 비중 예측 MAE로 신뢰도를 가늠한다.
추가로, 홀드아웃 잔차의 항목별 표준편차에 기반한 **95% 예측구간**(정규 근사)을 산출해
점추정 전망의 불확실성을 함께 제시한다.


```python
from sklearn.linear_model import LinearRegression
FUT = [("2026Q2", 17, 2), ("2026Q3", 18, 3)]   # (라벨, t, 분기)

def design(t_arr, q_arr):
    t = np.asarray(t_arr, float).reshape(-1, 1)
    qd = np.column_stack([(np.asarray(q_arr)==k).astype(float) for k in (2,3,4)])  # Q1 기준
    return np.hstack([t, qd])

preds = []
for qv in range(1, 6):
    sub = panel[panel["분위"]==qv].sort_values("t")
    Xtr = design(sub["t"], sub["분기"])
    fc = {}
    for k in CATS:
        m = LinearRegression().fit(Xtr, sub[f"비중_{k}"].values)
        fc[k] = np.clip(m.predict(design([t for _,t,_ in FUT], [q for *_,q in FUT])), 0, None)
    F = pd.DataFrame(fc, index=[lab for lab,_,_ in FUT])
    F = F.div(F.sum(axis=1), axis=0); F["분위"] = qv
    preds.append(F)
pred = pd.concat(preds)
print("[2026Q2/Q3 예측 비중(%) — 분위1 vs 분위5]")
for qv in (1, 5):
    cols = ["식료품·비주류음료","주거·수도·광열","교육","오락·문화","교통·운송"]
    print(f"\n 분위{qv}:"); print((pred[pred["분위"]==qv][cols]*100).round(1).to_string())

# 홀드아웃 검증(MAE)
val = ["2025Q3","2025Q4","2026Q1"]; errs = []
for qv in range(1, 6):
    sub = panel[panel["분위"]==qv].sort_values("t")
    tr = sub[~sub["분기라벨"].isin(val)]; va = sub[sub["분기라벨"].isin(val)]
    for k in CATS:
        m = LinearRegression().fit(design(tr["t"],tr["분기"]), tr[f"비중_{k}"].values)
        p = np.clip(m.predict(design(va["t"],va["분기"])), 0, None)
        errs.append(np.abs(p - va[f"비중_{k}"].values))
print(f"\n[홀드아웃(2025Q3~2026Q1) 비중 예측 MAE: {np.concatenate(errs).mean()*100:.2f}%p]")

# 95% 예측구간: 홀드아웃 잔차의 항목별 표준편차 기반 ±1.96σ (정규 근사)
res_by_item = {k: [] for k in CATS}
for qv in range(1, 6):
    sub = panel[panel["분위"]==qv].sort_values("t")
    tr = sub[~sub["분기라벨"].isin(val)]; va = sub[sub["분기라벨"].isin(val)]
    for k in CATS:
        m = LinearRegression().fit(design(tr["t"],tr["분기"]), tr[f"비중_{k}"].values)
        p = np.clip(m.predict(design(va["t"],va["분기"])), 0, None)
        res_by_item[k].extend((va[f"비중_{k}"].values - p).tolist())
sigma = {k: np.std(v, ddof=1) for k, v in res_by_item.items()}

print("\n[2026Q2 예측 95% 구간(%) — 분위1·분위5, 주요 항목 · 비중≥0이므로 하한 0 클립]")
for qv in (1, 5):
    pq = pred[pred["분위"]==qv]
    print(f" 분위{qv}:")
    for k in ["식료품·비주류음료","주거·수도·광열","교육","오락·문화"]:
        c = float(pq.loc["2026Q2", k])
        lo = max(0.0, c - 1.96 * sigma[k]); hi = c + 1.96 * sigma[k]
        print(f"   {k}: {c*100:.1f}%  [{lo*100:.1f}, {hi*100:.1f}]")
```

### 3-e. 강건성 점검 — 체감물가 지수 공식(파셰·피셔)과 분위별 디플레이터

라스파이레스(2022Q1 바스켓 고정)는 기준시점 선택에 의존한다. 누적 체감물가의 역진성이 공식에 강건한지
**파셰(비교시점 바스켓)·피셔(기하평균)** 지수로 교차 확인한다. 또한 방법1 디플레이터의 정보통신·오락문화
하위 가중치를 전체 평균이 아닌 **분위별 지출 가중치**로 재구성해 역진성을 재계산한다(품목별 CPI 파일이
있을 때만 수행). 12대 항목 수준에서만 바스켓을 차별화하므로 항목 *내부* 품목 구성의 계층차는 반영되지
않아, 측정된 역진성은 **하한**일 수 있다.


```python
# (1) 지수 공식 민감도: 라스파이레스 vs 파셰 vs 피셔 (2022Q1→2026Q1 누적 체감물가)
W22 = panel[panel["분기라벨"]=="2022Q1"].set_index("분위")[[f"비중_{k}" for k in CATS]]
W26 = panel[panel["분기라벨"]=="2026Q1"].set_index("분위")[[f"비중_{k}" for k in CATS]]
P22 = defl.loc["2022Q1", CATS].values; P26 = defl.loc["2026Q1", CATS].values

rows = []
for q in range(1, 6):
    w22 = W22.loc[q].values; w26 = W26.loc[q].values
    L = np.dot(w22, P26) / np.dot(w22, P22)        # Laspeyres
    P = np.dot(w26, P26) / np.dot(w26, P22)        # Paasche
    F = (L * P) ** 0.5                              # Fisher
    rows.append({"분위": q, "라스파이레스(%)": (L-1)*100, "파셰(%)": (P-1)*100, "피셔(%)": (F-1)*100})
idx_cmp = pd.DataFrame(rows).set_index("분위").round(2)
print("[누적 체감물가 2022Q1→2026Q1 — 지수 공식 비교]")
print(idx_cmp.to_string())
for c in idx_cmp.columns:
    print(f"  역진성(분위1-분위5) {c}: {idx_cmp.loc[1, c] - idx_cmp.loc[5, c]:+.2f}%p")

# (2) 분위별 하위 가중치 디플레이터 (품목별 CPI 파일이 있을 때만 — 2-d의 매핑 재사용)
item_fp = glob.glob(os.path.join(UPLOAD_DIR, "품목별_소비자물가지수*"))
if item_fp:
    icpi = pd.read_excel(item_fp[0], sheet_name="데이터", header=0)
    icpi.columns = ["시도", "품목"] + list(icpi.columns[2:])
    icpi = icpi.drop(columns=["시도"]).dropna(subset=["품목"]).reset_index(drop=True)
    icpi["name"] = (icpi["품목"].astype(str).str.replace("　","",regex=False)
                    .str.replace(" ","",regex=False).str.strip("()"))
    qc = [c for c in icpi.columns if c not in ("품목","name")]
    icpi = icpi.rename(columns={c: q_label(c) for c in qc})
    cidx = icpi.set_index("name")[[q_label(c) for c in qc]]

    def sub_price(items):
        if not items: return None
        return pd.concat([cidx.loc[n].astype(float) for n in items], axis=1).mean(axis=1)

    def composite_q(mapping, cat, dq):
        s, wt = {}, {}
        for sub, items in mapping.items():
            sp = sub_price(items)
            if sp is None: continue
            s[sub] = sp
            wt[sub] = np.average(dq[f"{PRE}{cat}_{sub}"], weights=dq["가중값"])
        wser = pd.Series(wt); wser /= wser.sum()
        return pd.DataFrame(s).T.mul(wser, axis=0).sum(axis=0)

    df_raw = pd.read_pickle(os.path.join(WORK_DIR, "df_raw.pkl"))
    cum_q = {}
    for q in range(1, 6):
        dq = df_raw[df_raw["분위"] == q]
        dfl_q = defl.copy()
        dfl_q["정보통신"]  = composite_q(ICT_MAP, "정보통신", dq)
        dfl_q["오락·문화"] = composite_q(REC_MAP, "오락문화", dq)
        w22 = W22.loc[q].values
        cum_q[q] = (np.dot(w22, dfl_q.loc["2026Q1", CATS].values) /
                    np.dot(w22, dfl_q.loc["2022Q1", CATS].values) - 1) * 100
    print("\n[분위별 하위 가중치 디플레이터 기반 누적 체감물가(%) — 라스파이레스]")
    for q, v in cum_q.items(): print(f"  분위{q}: {v:.2f}%")
    print(f"  역진성(분위1-분위5): {cum_q[1] - cum_q[5]:+.2f}%p  (전체 평균 가중치 대비 비교)")
else:
    print("\n품목별 CPI 파일 없음 → 분위별 디플레이터 강건성 점검 생략")
```

### 3-f. 핵심 결과의 통계적 불확실성 — 가구 부트스트랩 신뢰구간

체감물가 역진성과 구조격차지수 변화는 점추정만으로는 표본 변동 대비 의미 있는지 알 수 없다.
**분기×분위 층화 가구 리샘플링(B=500)** 으로 95% 신뢰구간을 산출한다.
구간이 0을 포함하지 않으면 결론(역진성·고착)이 통계적으로 지지되고, 포함하면 그만큼 보수적으로 서술한다.


```python
rng = np.random.default_rng(0)
B = 500
df_raw = pd.read_pickle(os.path.join(WORK_DIR, "df_raw.pkl"))
use_q = ["2022Q1", "2026Q1"]
cat_cols = list(CAT12.values())
cells_b = {}
for (ql, q), g in df_raw[df_raw["분기라벨"].isin(use_q)].groupby(["분기라벨", "분위"]):
    cells_b[(ql, int(q))] = (g[cat_cols].to_numpy(float), g["가중값"].to_numpy(float))

P22 = defl.loc["2022Q1", CATS].values; P26 = defl.loc["2026Q1", CATS].values

def boot_shares(M, w, idx):
    Mw = (M[idx] * w[idx, None]).sum(axis=0)
    return Mw / Mw.sum()

regr, d22, d26, dchg = [], [], [], []
for b in range(B):
    sh = {}
    for key, (M, w) in cells_b.items():
        n = len(w); idx = rng.integers(0, n, n)
        sh[key] = boot_shares(M, w, idx)
    cum = {q: np.dot(sh[("2022Q1", q)], P26) / np.dot(sh[("2022Q1", q)], P22) for q in (1, 5)}
    regr.append((cum[1] - cum[5]) * 100)
    g22 = 0.5 * np.abs(sh[("2022Q1", 5)] - sh[("2022Q1", 1)]).sum()
    g26 = 0.5 * np.abs(sh[("2026Q1", 5)] - sh[("2026Q1", 1)]).sum()
    d22.append(g22); d26.append(g26); dchg.append(g26 - g22)

def ci(a):
    lo, hi = np.percentile(np.asarray(a), [2.5, 97.5]); return lo, hi

print(f"[가구 부트스트랩 B={B} · 분기×분위 층화 리샘플링 — 95% CI]")
lo, hi = ci(regr)
print(f"  체감물가 역진성(분위1-분위5, %p): [{lo:.2f}, {hi:.2f}]"
      f"  → 0 미포함이면 역진성 통계적 지지")
lo, hi = ci(d22); print(f"  구조격차지수 2022Q1: [{lo:.3f}, {hi:.3f}]")
lo, hi = ci(d26); print(f"  구조격차지수 2026Q1: [{lo:.3f}, {hi:.3f}]")
lo, hi = ci(dchg)
print(f"  구조격차지수 변화(26Q1-22Q1): [{lo:.3f}, {hi:.3f}]"
      f"  → 0 포함 시 '축소'가 아닌 '고착'으로 서술")
```

### 4-a. 9종 알고리즘 학습 및 성능 비교

거리·커널·신경망 계열(SVR·MLP)은 표준화 파이프라인으로 학습한다.
검증 MAE 기준 오름차순으로 정렬하고, 과적합/부진 여부를 자동 판정한다.

```python
models = {
    "LinearRegression": LinearRegression(),
    "Ridge":            Ridge(alpha=1.0),
    "RandomForest":     RandomForestRegressor(n_estimators=400, random_state=0),
    "ExtraTrees":       ExtraTreesRegressor(n_estimators=400, random_state=0),
    "HistGradBoost":    HistGradientBoostingRegressor(random_state=0),
    "XGBoost":          XGBRegressor(n_estimators=400, learning_rate=0.05, max_depth=3,
                                     random_state=0, verbosity=0),
    "LightGBM":         LGBMRegressor(n_estimators=400, learning_rate=0.05, max_depth=3,
                                      random_state=0, verbose=-1),
    "SVR(RBF)":         make_pipeline(StandardScaler(), SVR(C=1.0, kernel="rbf")),
    "MLP":              make_pipeline(StandardScaler(),
                                      MLPRegressor(hidden_layer_sizes=(64, 32),
                                                   max_iter=2000, random_state=0)),
}

rows, fitted = [], {}
for name, mdl in models.items():
    mdl.fit(X_train, Y_train)
    fitted[name] = mdl
    p_tr, p_va = mdl.predict(X_train), mdl.predict(X_val)
    rows.append({
        "모델": name,
        "학습MAE": mean_absolute_error(Y_train, p_tr) * 100,
        "검증MAE": mean_absolute_error(Y_val,   p_va) * 100,
        "검증RMSE": mean_squared_error(Y_val,   p_va) ** 0.5 * 100,
        "학습R2": r2_score(Y_train, p_tr),
        "검증R2": r2_score(Y_val,   p_va),
    })

res = pd.DataFrame(rows).sort_values("검증MAE").reset_index(drop=True)
res["R2격차"] = res["학습R2"] - res["검증R2"]

def judge(r):
    if r["검증R2"] < 0.5:                       return "부진(과소적합/저성능)"
    if r["R2격차"] > 0.10 or r["학습MAE"] < 0.05: return "과적합 주의"
    return "양호"
res["판정"] = res.apply(judge, axis=1)

pd.set_option("display.float_format", lambda x: f"{x:.3f}")
print("[알고리즘별 성능 비교 — MAE/RMSE 단위 %p]")
print(res.to_string(index=False))

over = res[res["판정"] == "과적합 주의"]["모델"].tolist()
poor = res[res["판정"].str.startswith("부진")]["모델"].tolist()
print("\n※ 과적합 주의:", over if over else "없음")
print("※ 검증 부진  :", poor if poor else "없음")
```

### 4-b. 베이스라인 대비 성능·항목별 오차·확장 윈도우 CV

높은 검증 R²는 항목 더미가 비중의 '수준 차이'(식료품 ~20% vs 주류담배 ~1.5%)를 설명하는 데서 상당 부분
나온다. 따라서 ① **naive 베이스라인 2종**(항목×분위 학습기간 평균 / 직전 분기 값 persistence)과 비교해
모델의 *실질 개선폭*을 확인하고, ② **항목별 MAE**로 비중이 작은 항목의 상대오차를 점검하며,
③ 단일 홀드아웃의 한계를 보완하는 **확장 윈도우(expanding-window) 시계열 CV**로 1분기 선행 예측 성능을
검증한다. (4-a 셀 실행 후 같은 세션에서 수행)


```python
from sklearn.metrics import mean_absolute_error
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor

panel = pd.read_pickle(os.path.join(WORK_DIR, "panel.pkl"))
CATS = [c[3:] for c in panel.columns if c.startswith("비중_")]
long = panel.melt(id_vars=["분기라벨","분위","t","분기"],
                  value_vars=[f"비중_{k}" for k in CATS],
                  var_name="항목", value_name="비중")
long["항목"] = long["항목"].str[3:]
val_q = ["2025Q3","2025Q4","2026Q1"]
tr, va = long[~long["분기라벨"].isin(val_q)], long[long["분기라벨"].isin(val_q)].copy()

# ① naive 베이스라인: (a) 항목×분위 학습기간 평균  (b) 직전 관측 분기(2025Q2) 값 고정
mean_map = tr.groupby(["분위","항목"])["비중"].mean()
va["pred_mean"] = [mean_map.get(k) for k in zip(va["분위"], va["항목"])]
last = long[long["분기라벨"]=="2025Q2"].set_index(["분위","항목"])["비중"]
va["pred_last"] = [last.get(k) for k in zip(va["분위"], va["항목"])]
mae_mean = mean_absolute_error(va["비중"], va["pred_mean"]) * 100
mae_last = mean_absolute_error(va["비중"], va["pred_last"]) * 100
print(f"[naive 베이스라인 검증 MAE(%p)]  항목×분위 평균: {mae_mean:.3f}  /  persistence(2025Q2): {mae_last:.3f}")
print("→ 4-a의 모델 MAE는 이 값 '대비 개선폭'으로 해석해야 함")

# ② 항목별 검증 MAE — 대표 모델 RandomForest (4-a의 fitted 사용)
rf = fitted["RandomForest"]
va = va.reset_index(drop=True)
va["pred_rf"] = rf.predict(X_val)
item_mae = (va.assign(err=(va["비중"]-va["pred_rf"]).abs()*100)
              .groupby("항목")["err"].mean().sort_values(ascending=False))
print("\n[항목별 검증 MAE(%p) — RandomForest]")
print(item_mae.round(3).to_string())
print("※ 비중이 작은 항목(주류·담배 등)은 MAE가 작아도 상대오차가 클 수 있음")

# ③ 확장 윈도우 CV: 각 분기를 '그 이전 데이터로만' 학습해 1분기 선행 예측
Xall = pd.get_dummies(long[["t","분기","분위","항목"]].astype({"분기":str,"분위":str}),
                      columns=["분기","분위","항목"], drop_first=True).astype(float)
Yall = long["비중"].astype(float)
folds = ["2024Q3","2024Q4","2025Q1","2025Q2","2025Q3","2025Q4","2026Q1"]
cv_models = {"Linear":        LinearRegression(),
             "RandomForest":  RandomForestRegressor(n_estimators=200, random_state=0),
             "HistGradBoost": HistGradientBoostingRegressor(random_state=0)}
t_of = dict(zip(long["분기라벨"], long["t"]))
cv = {nm: [] for nm in list(cv_models) + ["Naive(평균)", "Persistence"]}
for ql in folds:
    tr_m = (long["t"] < t_of[ql]).values; te_m = (long["분기라벨"] == ql).values
    for nm, mdl in cv_models.items():
        mdl.fit(Xall[tr_m], Yall[tr_m])
        cv[nm].append(mean_absolute_error(Yall[te_m], mdl.predict(Xall[te_m])) * 100)
    # naive 베이스라인도 동일 방식(과거 데이터만 사용)으로 평가
    te_long = long[te_m]
    mmap = long[tr_m].groupby(["분위","항목"])["비중"].mean()
    p_mean = [mmap.get(k) for k in zip(te_long["분위"], te_long["항목"])]
    prev = long[long["t"] == t_of[ql] - 1].set_index(["분위","항목"])["비중"]
    p_last = [prev.get(k) for k in zip(te_long["분위"], te_long["항목"])]
    cv["Naive(평균)"].append(mean_absolute_error(te_long["비중"], p_mean) * 100)
    cv["Persistence"].append(mean_absolute_error(te_long["비중"], p_last) * 100)
print("\n[확장 윈도우 CV — 1분기 선행 예측 MAE(%p), 7개 분기 평균±표준편차, 오름차순]")
for nm, e in sorted(cv.items(), key=lambda kv: np.mean(kv[1])):
    print(f"  {nm:14s}: {np.mean(e):.3f} ± {np.std(e):.3f}")
```

### 4-c. 결과 해석

- **naive 베이스라인이 강력하다.** 항목×분위 학습기간 평균만으로 검증 MAE 0.611%p(persistence 0.652%p).
  비중 분산의 대부분이 '항목×계층의 평균 수준'으로 설명된다는 뜻이며, 검증 R²≈0.94~0.99를 그대로
  모델 성능으로 읽으면 안 되는 이유다.
- **naive 평균을 능가하는 모델은 RandomForest·ExtraTrees뿐.** RandomForest 검증 MAE 0.454%p로
  naive 평균 대비 약 26% 개선, 확장 윈도우 CV에서도 가장 낮고 안정적. 반면
  HistGradientBoosting(0.743)·XGBoost(0.883)·LightGBM(1.410)·선형(≈1.5)은 naive 평균보다도 오차가
  크다 — 가산적(additive)·얕은 구조로는 항목×분위 상호작용(같은 항목이라도 분위마다 다른 수준·추세)을
  충분히 포착하지 못하기 때문이다.
- **과적합 주의(ExtraTrees).** 학습 MAE≈0(완전 암기). 검증 MAE는 근소하게 최저(0.436)이나,
  일반화 신뢰를 위해 대표 모델로는 RandomForest를 채택한다.
- **검증 부진(SVR·MLP).** 소표본·더미 위주 설계에서 커널·신경망은 본 과제에 부적합(검증 R²<0.5).
- **결론.** 모델의 실질 가치는 '평균 수준' 위에서의 추세·계절 미세조정이며, 그 개선폭
  (naive 대비 약 0.16%p)이 단기 전망(3-d)의 신뢰 근거가 된다.

### 5-c. 해석 — 학습 재현 근거 vs 검증 활용 근거

**학습 재현의 패턴 근거(그림 1).** 모델은 소비구조 비중을 *항목 정체성*으로 가장 크게 재현한다 —
식료품·비주류음료(≈25%)·음식숙박(≈20%)·주거수도광열(≈17%)이 비중 '수준'을 결정한다.
이는 항목마다 기준 비중이 크게 다르다는 사실(식료품 ~20% vs 주류담배 ~1.5%)을 학습한 결과다.
그다음 *소득분위*(분위5·4 더미)가 그 비중을 계층별로 재조정하고, 추세(t)·분기(계절)가 미세 조정한다.

**검증 활용의 패턴 근거(그림 2·3).** 미래 분기(2025Q3~2026Q1) 예측에서도 동일하게
**항목 → 소득분위** 순서가 지배하며, 특히 **분위5**의 기여(순열·SHAP 모두 상위)가 뚜렷하다 —
고소득층의 구조가 가장 구별되기 때문이다. 학습에서 중요했던 변수가 검증에서도 그대로 중요하다는 것은,
모델이 *허위 신호가 아니라 안정적인 '항목×계층' 구조*에 의존함을 의미하며, 단기 예측의 신뢰 근거가 된다.

**요약.** 소비구조는 "무엇을 사느냐(항목)"가 골격을 정하고 "누가 사느냐(소득분위)"가 형태를 바꾸며,
시간은 그 위에서 완만한 추세·계절로 작동한다 — 본 분석의 계층별 격차 서사와 정확히 일치한다.

**한계 — 외삽.** 트리 계열 모델은 학습 범위 밖의 t를 외삽하지 못한다(미래 t는 마지막 학습 t와 동일하게 취급).
따라서 본 절의 모델은 *구조 재현·해석용*이며, 미래 분기 '전망'은 3-d의 추세+계절 회귀가 담당한다 —
두 역할을 혼동하지 않는다.


## 6. 현실 적용

분석 결과로 *"소득 5분위 계층별 소비구조 변화·격차"* 라는 정의된 문제를 얼마나 풀었는지 평가하고,
정책·현실에 적용 가능한 시나리오를 정리한다. 먼저 핵심 지표를 한 화면에 종합한다.

```python
import warnings, numpy as np, pandas as pd; warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt
set_korean_font()  # 0단계 정의 — OS 독립 한글 폰트 설정

panel = pd.read_pickle(os.path.join(WORK_DIR, "panel.pkl"))
QORDER = sorted(panel["분기라벨"].unique())
def at(q, ql, col): return panel[(panel.분위==q)&(panel.분기라벨==ql)][col].values[0]

# 핵심 지표 종합표
rows = []
for q in range(1, 6):
    rows.append({"분위": q,
        "체감물가누적(%)": round((at(q,"2026Q1","체감물가")/at(q,"2022Q1","체감물가")-1)*100, 2),
        "엥겔 22Q1→26Q1(%)": f"{at(q,'2022Q1','엥겔계수')*100:.1f}→{at(q,'2026Q1','엥겔계수')*100:.1f}",
        "슈바베 22Q1→26Q1(%)": f"{at(q,'2022Q1','슈바베계수')*100:.1f}→{at(q,'2026Q1','슈바베계수')*100:.1f}"})
summary = pd.DataFrame(rows).set_index("분위")
print("[표. 핵심 지표 종합 — 분위별]")
print(summary.to_string())
print("\n체감물가 역진성(분위1 - 분위5):",
      round(summary.loc[1,"체감물가누적(%)"]-summary.loc[5,"체감물가누적(%)"], 2), "%p")

# 종합 그림: 엥겔 추이 + 체감물가 누적
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
for q in range(1, 6):
    s = panel[panel.분위==q].set_index("분기라벨").loc[QORDER]
    ax[0].plot(QORDER, s["엥겔계수"]*100, marker="o", ms=3, label=f"분위{q}")
ax[0].set_title("그림. 엥겔계수 추이(분위별)"); ax[0].set_ylabel("%")
ax[0].tick_params(axis="x", rotation=90); ax[0].legend(fontsize=8)
cum = [(at(q,"2026Q1","체감물가")/at(q,"2022Q1","체감물가")-1)*100 for q in range(1,6)]
ax[1].bar([f"분위{q}" for q in range(1,6)], cum, color="#E45756"); ax[1].set_ylim(12, 13.2)
ax[1].set_title("그림. 체감물가 누적상승률(2022Q1→2026Q1)"); ax[1].set_ylabel("%")
plt.tight_layout(); plt.show()
```

### 6-a. 문제 해결도 평가

정의한 문제 — *"계층별 소비구조가 어떻게 변했고 격차가 어떻게 움직였는가"* — 에 대해 다음을 규명했다.

1. **소비구조와 격차 정량화**: 17개 분기×5분위 패널로 12대 항목 비중을 산출. 저소득층은 필수재
   (식료품·주거·보건)에, 고소득층은 선택재(교육·교통·여가)에 집중. 구조격차지수는 0.262→0.257
   (중간 2024Q1에 0.303까지 상승 후 회귀)이며, 변화량의 부트스트랩 95% CI [−0.042, +0.038]가
   0을 포함해 **확대도 축소도 아닌 구조적 고착**이라는 결론이 통계적으로 뒷받침된다.
2. **인플레이션의 역진성**: 분류 정합 디플레이터(방법1) 기준 누적 체감물가는 분위1 12.98% >
   분위5 12.51%(+0.47%p). 부트스트랩 95% CI [+0.14, +0.78]가 0을 배제하고, 파셰(+0.82%p)·
   피셔(+0.64%p)·분위별 하위 가중치 디플레이터(+0.40%p)에서도 부호가 유지되어 **지수 공식과
   표본 변동에 강건**하다. 항목 내 품목 구성의 계층차는 미반영이므로 실제 역진성의 **하한**일 수 있다.
3. **후생 지표**: 엥겔계수(분위1 22.9→20.8% vs 분위5 14.4→11.6%)·슈바베계수(분위1 22.7→21.7% vs
   분위5 10.8→10.4%)로 저소득층의 식료품·주거 부담이 약 2배임을 확인. 엥겔계수는 전반적으로
   하향(소득 회복)하나 격차는 지속된다.
4. **계층별 행태**: 식료품 지출탄력성은 대체로 e<1로 필수재 성격이 지지된다(분위2만 CI가 1 포함).
   오락·문화는 분위4·5에서 CI 하한이 1을 넘어 사치재 성격이 통계적으로 지지되고, 교육은 점추정
   e>1이나 모든 분위에서 CI가 1을 포함해 단정할 수 없다. 식료품가 1%↑ 시 비식료품 실질지출
   반응 η는 분위1에서 음(−0.17)이지만 CI가 0을 포함해 '긴축 방향'을 단정할 수 없으며, 유일하게
   유의한 분위5(+2.48, CI [+1.15, +3.81])는 고소득층의 물가 충격 흡수 여력을 시사한다
   (식별이 약한 탐색적 결과).
5. **예측 가능성**: 비중 분산의 대부분은 항목×분위 평균(naive 검증 MAE 0.611%p)으로 설명되며,
   RandomForest가 이를 약 26% 개선(검증 0.454%p, 확장 윈도우 CV에서도 최저)해 2026Q2/Q3
   단기 전망의 근거가 된다. XAI 결과 예측은 *항목→소득분위*의 안정적 구조에 의존(허위 신호
   아님)함을 확인했다.

→ **정의된 문제를 정량적으로, 그리고 불확실성 정량화와 함께 해결**했다. 격차의 *수준·고착*은
역진성과 함께 통계적으로 지지되나 *급격한 확대*는 관찰되지 않았으며, 인과가 아닌 기술적
진단이라는 점은 분명히 한다.


### 6-b. 정책·현실 적용 시나리오

- **타깃형 물가 대응**: 체감물가 부담이 뚜렷이 높은 분위1에 식료품·에너지 바우처를 우선 배분
  (분위2~4의 누적 체감물가는 12.62~12.70%로 격차가 0.1%p 이내라 분위1 집중이 효율적).
  *분위별 체감물가* 를 분기 모니터링 지표로 상시 산출해 지원 규모·시점을 조정.
- **주거비(슈바베) 안정**: 저소득층 주거비 부담이 약 22%로 고착 → 전·월세 지원, 공공임대 확대,
  에너지 바우처를 통해 슈바베계수 상승을 억제.
- **소비 위축 조기경보 대시보드**: `분기×분위×항목`의 비중·실질지출·격차지수를 분기 갱신해,
  저소득층의 **보건·교육 지출 감소** 같은 고통스러운 긴축 신호를 선제적으로 포착.
- **단기 전망 활용**: 2026Q2/Q3 소비구조 전망으로 복지 예산·농축수산 수급을 선제 편성.
  기업은 계층별 소비구조 전망을 세그먼트 마케팅·수요예측 입력으로 활용.
- **공식 통계 보완·연계**: 통계청 *가구특성별 물가지수(실험적통계)* 와 교차검증하고,
  방법1 분류정합 디플레이터로 2025 분류개편 전후 시계열의 연속성·비교가능성을 확보.
